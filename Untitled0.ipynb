```python
def outlier_calculator(col):
    Q1 = X_train_temp[col].quantile(0.25)
    Q3 = X_train_temp[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = X_train_temp[(X_train_temp[col] < lower) | (X_train_temp[col] > upper)]
    print(f"Number of detected outliers in {col}: {len(outliers)}")
```

## Pipeline Creation

```python
# Log Transform
numeric_log_median_features = ['ApplicantIncome', 'CoapplicantIncome', 'LoanAmount']

# def log_transform(x):
#     return np.log(x + 1)

# log_transformer = FunctionTransformer(log_transform)
log_transformer = FunctionTransformer(np.log1p)

num1_transformer = Pipeline(
    steps=[
        ('imputer', SimpleImputer(strategy='median')),
        ('log_transform', log_transformer),
        ('scaling', StandardScaler())
    ]
)
```

```python
numeric_mode1_features = ['Dependents', 'Loan_Amount_Term', 'FamilySize']
num2_transformer = Pipeline(
    steps=[
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('scaling', StandardScaler())
    ]
)
```

```python
numeric_mode2_features = ['Credit_History']
num3_transformer = Pipeline(
    steps=[
        ('imputer', SimpleImputer(strategy='most_frequent')),
    ]
)
```

```python
categorical_features = ['Gender', 'Married', 'Education', 'Self_Employed', 'Property_Area']
categorical_transformer = Pipeline(
    steps=[
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('encoder', OneHotEncoder(drop='if_binary'))
    ]
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('num1', num1_transformer, numeric_log_median_features),
        ('num2', num2_transformer, numeric_mode1_features),
        ('num3', num3_transformer, numeric_mode2_features),
        ('cat', categorical_transformer, categorical_features),
    ]
)
```

# Ensemble - boosting, stacking

```python
'''
As the primary base models, Logistic Regression, Random Forest and Gradient Boosting were selected. 

Logistic Regression was choosen as it was simple, works well with binary classification by employing sigmoid function where the probability of each of the class
is plugged in to get either 0 or 1. It further more helps us to understand how complex the relation is between features and target and if more complex or how much
more complex model is required.

Random forest classifier combines multiple decision trees using something called bagging. It handles non-linear relationships better. It is less prone to or
less effected by outliers. Even though trees in general has a tendency to overfit especially on smaller datasets, but by using multiple decision trees
that chance is also reduced. 

This is another ensemble algo where models are built by corrections erros that took in previous ite. It is known to provide good predictive performance, effective
for tabular datasets like ours and also handles non-linear relations between features and target effectively. 
'''
```

```python
#base learner
lr = LogisticRegression()
rf = RandomForestClassifier( n_estimators=100, random_state=42 )
gb = GradientBoostingClassifier( n_estimators=100 , random_state=42 )
```

```python
#Voting regressor
voting_classifer = VotingClassifier(
    estimators= [
        ('lr', lr),
        ('rf',rf),
        ('gb', gb)
    ]
)
```

```python
#stacking
stacking_Classifier = StackingClassifier(
    estimators= [
        ('rf', rf),
        ('gb', gb)
    ],
    final_estimator= RidgeClassifier() #the meta learner
)
```

# Model training

```python
#dictionary of all model
model_to_train = {
    'Logistic Regression' : lr,
    'Random Forest Classifier' : rf,
    'Gradient Boosting Classifier': gb,
    'Voting Ensemble ' : voting_classifer,
    'Stacking Ensemble ' : stacking_Classifier
}
```

```python
#training
result = []

for name , model in model_to_train.items():
  #create full pipeline with preprocessor
  pipe = Pipeline(
      [
          ('preprocessor', preprocessor),
          ('model',model)
      ]
  )

  #train
  pipe.fit(X_train,y_train)

  #predict
  y_pred = pipe.predict(X_test)

  #Evaluate
  acc = accuracy_score(y_test,y_pred)
  precision = precision_score(y_test,y_pred)
  recall = recall_score(y_test,y_pred)
  f1 = f1_score(y_test,y_pred)

  result.append({
      "Model": name,
      "Accuracy" :acc,
      "Precision": precision,
      "Recall" : recall,
      "F1 Score" : f1,
  })
```

```python
results_df = pd.DataFrame(result).sort_values("F1 Score", ascending=False)
results_df
```

# Cross Validation

```python
lr_pipeline = Pipeline(
    steps=[
        ('preprocessor',preprocessor),
        ('model',LogisticRegression())
     ]
  )
```

```python
# 10 fold cv
skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)
cv_scores = cross_val_score(
    lr_pipeline,
    X_train,
    y_train,
    cv=skf,
    n_jobs=-1,
    scoring='f1',
)
```

```python
print(f"F1 Score for each fold\n: {cv_scores}\n")
print(f"Mean F1 Score: {cv_scores.mean()}\n")
print(f"Standard Deviation: {cv_scores.std()}")
```

# Grid Search CV

```python
param_grid = {
    'model__C': [0.01, 0.1, 1, 10, 100],
    'model__penalty': ['l2'],
    'model__solver': ['lbfgs', 'liblinear'],
    'model__max_iter': [100, 200, 500]
}
```

```python
grid_search = GridSearchCV(
    estimator = lr_pipeline,
    param_grid = param_grid,
    cv = skf,
    scoring = 'f1',
    n_jobs =-1,
    verbose = 2
)

grid_search.fit(X_train,y_train)
```

```python
print(grid_search.best_score_)
```
